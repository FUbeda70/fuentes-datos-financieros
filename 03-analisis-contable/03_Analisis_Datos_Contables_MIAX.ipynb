Descomponemos la rotación
$$\frac{Ventas}{\text{Activo Total}}=\left\{
\begin{array}{l}
\frac{Ventas}{\text{Activo Corriente}} \\
\frac{Ventas}{\text{Activo No Corriente}}\end{array}
\right.$$

```python
def obtener_datos(ticker):
    empresa = yf.Ticker(ticker)
    resultados = empresa.income_stmt
    balance = empresa.balance_sheet
    cashflow = empresa.cashflow

    ventas = resultados.loc["Total Revenue"].iloc[0]
    EBIT = resultados.loc["Operating Income"].iloc[0]
    activo_total=balance.loc["Total Assets"].iloc[0]
    activo_corriente = balance.loc["Current Assets"].iloc[0]
    activo_no_corriente = balance.loc["Total Non Current Assets"].iloc[0]


    return ventas, EBIT, activo_total, activo_corriente, activo_no_corriente
```

# ROE

## Rentabilidad Financiera
$$ROE=\frac{\text{Beneficio Neto}}{Patrimonio}$$

## Apalancamiento Financiero

$$ROC=\frac{\text{Beneficio sin Intereses y Después de Impuestos}}{Activo Total}$$

$$ROE=ROC+\left(ROC-k_d(1-\tau)\times \frac{Deuda}{Patrimonio}\right)$$

$$k_d=\frac{Intereses}{Deuda}$$


$$\tau=\frac{Impuestos}{Beneficio Neto}$$

```python
def obtener_datos_deuda(ticker):
    empresa = yf.Ticker(ticker)
    resultados = empresa.income_stmt
    balance = empresa.balance_sheet
    cashflow = empresa.cashflow


    EBIT = resultados.loc["Operating Income"].iloc[0]
    beneficio_neto = resultados.loc["Net Income"].iloc[0]
    intereses = resultados.loc["Interest Expense"].iloc[0]
    impuestos= resultados.loc["Tax Provision"].iloc[0]
    patrimonio_neto = balance.loc["Stockholders Equity"].iloc[0]
    activo_total=balance.loc["Total Assets"].iloc[0]
    deuda_total = balance.loc["Total Debt"].iloc[0]

    return EBIT, beneficio_neto, patrimonio_neto, activo_total, deuda_total, intereses, impuestos
```

```python
datos=[]
for ticker in tickers:
    try:
        EBIT, beneficio_neto, patrimonio_neto, activo_total, deuda_total,intereses, impuestos = obtener_datos_deuda(ticker)

        datos.append([
            ticker,
            EBIT,
            beneficio_neto,
            patrimonio_neto,
            activo_total,
            deuda_total,
            intereses,
            impuestos
        ])
    except KeyError as e:
      print(f"Advertencia: No se pudieron obtener los datos de {ticker}. Falta la clave: {e}")

    except IndexError:
      print(f"Advertencia: No se pudieron obtener los datos de {ticker}. Los datos pueden estar vacíos o incompletos.")
```

```python
datos
```

```python
df_deuda = pd.DataFrame(
    datos,
    columns=[
        "Ticker",
        "EBIT",
        "Beneficio_Neto",
        "Patrimonio_Neto",
        "Activo_Total",
        "Deuda_Total",
        "Intereses",
        "Impuestos"
    ]
)

df_deuda
```

```python
def ratios_deuda(df):
    ratios=pd.DataFrame()

    ratios["Ticker"]=df["Ticker"]

    ratios["kd"]=df["Intereses"]/df["Deuda_Total"]
    ratios["tau"]=df["Impuestos"]/df["Beneficio_Neto"]


    ratios["ROC"] = (
    df["EBIT"]*(1-ratios["tau"]) / df["Activo_Total"]
    )
    ratios["ROE"] = (
        df["Beneficio_Neto"] / df["Patrimonio_Neto"]
    )
    ratios["Apalancamiento"] = (
        df["Deuda_Total"] / df["Patrimonio_Neto"]
    )
    return ratios
```

```python
ratios_deuda=ratios_deuda(df_deuda)
ratios_deuda
```

# RATIOS DE MERCADO

$$Price-to-book=\frac{\text{Capitalización Bursatil}}{\text{Patrimonio Neto}}$$


$$PER=\frac{Precio}{\text{Benefcio Neto}}$$




```python
empresa = yf.Ticker("ITX.MC")

price_to_book = empresa.info.get("priceToBook")
per = empresa.info.get("trailingPE")

print(price_to_book,per)
```

Compara el PER y Price to Book entre todas la empresas

```python
def obtener_datos_mercado (ticker):
    empresa = yf.Ticker(ticker)

    price_to_book = empresa.info.get("priceToBook")
    per = empresa.info.get("trailingPE")

    return price_to_book, per
```

```python
datos=[]
for ticker in tickers:
    try:
        price_to_book, per = obtener_datos_mercado (ticker)

        datos.append([
             ticker,
             price_to_book,
             per
        ])
    except KeyError as e:
      print(f"Advertencia: No se pudieron obtener los datos de {ticker}. Falta la clave: {e}")

    except IndexError:
      print(f"Advertencia: No se pudieron obtener los datos de {ticker}. Los datos pueden estar vacíos o incompletos.")
```
